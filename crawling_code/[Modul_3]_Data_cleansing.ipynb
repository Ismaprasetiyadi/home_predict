## Cleaning Kolom

### Harga Beli

```python
data['price'].head(10)
```

```python
def convert_price(price):
    if pd.isna(price):
        return None
    price = price.replace('Rp ', '').replace('.', '').replace(',', '.')
    try:
        if 'Jt' in price:
            return float(price.replace('Jt', '')) * 1_000_000
        elif 'M' in price:
            return float(price.replace('M', '')) * 1_000_000_000
        else:
            return float(price)
    except ValueError:
        return None
```

```python
data['Harga_Asli'] = data['price'].apply(convert_price)
data_price = data[['price','Harga_Asli']]
data_price.head(10)
```

### Lokasi

```python
data['kecamatan'] = data['location'].str.split(',', expand=True)[0]
data_lokasi = data[['location', 'kecamatan']]
data_lokasi.head(10)
```

### bedroom

```python
# Fill NaN values with 0 before converting to integer
data['bedroom'] = data['bedrooms'].fillna(0).astype(int)
data_bedroom = data[['bedroom']]
display(data_bedroom.head(10))
```

### Bathroom

```python
data['bathrooms'] = data['bathrooms'].fillna(0).astype(int)
data_bathroom = data[['bathrooms']]
data_bathroom.head(10)
```

### Toilet Tamu

```python
def extract_guest_toilet_count(guest_toilet):
    if pd.isna(guest_toilet):
        return 0
    try:
        return int(guest_toilet.split(' ')[0])
    except:
        return 0
```

```python
data['jumlah_guest_toilet'] = data['guest_toilet'].apply(extract_guest_toilet_count)
data_toilet_tamu = data[['guest_toilet', 'jumlah_guest_toilet']].head()
data_toilet_tamu.head()
```

### Fasilitas Properti

```python
import ast

def parse_facilities_list(facilities_str):
    if pd.isna(facilities_str):
        return "Tidak Ada"
    try:
        facilities_list = ast.literal_eval(facilities_str)
        return ", ".join(facilities_list)
    except:
        return ""
```

```python
data['list_facilities'] = data['facilities_property'].apply(parse_facilities_list)
data_fasilitas_properti = data[['facilities_property', 'list_facilities']]
data_fasilitas_properti.head()
```

### Jumlah Lantai

```python
data['lantai'] = data['place_features.Jumlah lantai'].fillna(0).astype(int)
data['lantai'] = data['lantai'].replace(0, 1)

data_lantai = data[['lantai']]
data_lantai.head(10)
```

### Area yang dapat digunakan

```python
def extract_area(area_str):
    if pd.isna(area_str):
        return None
    try:
        area_numeric_str = area_str.replace(' m²', '').replace(',', '.')
        return float(area_numeric_str)
    except ValueError:
        return None
```

```python
data['area_digunakan'] = data['place_features.Area yang dapat digunakan'].apply(extract_area).fillna(0).astype(int)
data_luas_bangunan = data[['place_features.Area yang dapat digunakan', 'area_digunakan']]
display(data_luas_bangunan.head(10))
```

### luas tanah

```python
def extract_land_area(area_str):
    if pd.isna(area_str):
        return None
    try:
        area_numeric_str = area_str.replace(' m²', '').replace(',', '.')
        return float(area_numeric_str)
    except ValueError:
        return None
```

```python
data['luas_tanah'] = data['place_features.Land area'].apply(extract_land_area).fillna(0).astype(int)
data_luas_bangunan = data['luas_tanah']
data_luas_bangunan.head(10)
```

# Mengoreksi nilai outlier

```python
columns_to_summarize = ['Harga_Asli', 'area_digunakan', 'luas_tanah']
display(data_final[columns_to_summarize].describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

columns_to_plot = ['Harga_Asli', 'area_digunakan', 'luas_tanah']

plt.figure(figsize=(15, 5))
for i, col in enumerate(columns_to_plot):
    plt.subplot(1, 3, i + 1)
    sns.histplot(data=data_final, x=col, kde=True)
    plt.title(f'Distribusi dari {col}')
    plt.xlabel(col)
    plt.ylabel('Frekuensi')

plt.tight_layout()
plt.show()
```

```python
def detect_outliers_iqr(df, column, threshold=1.5):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - threshold * IQR
    upper_bound = Q3 + threshold * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers

# Deteski outlier
outliers_harga = detect_outliers_iqr(data_final, 'Harga_Asli')
outliers_area_digunakan = detect_outliers_iqr(data_final, 'area_digunakan')
outliers_luas_tanah = detect_outliers_iqr(data_final, 'luas_tanah')

# Jumlah outlier setiap kolom
print(f"Jumlah outlier di kolom 'Harga_Asli': {len(outliers_harga)}")
print(f"Jumlah outlier di kolom 'area_digunakan': {len(outliers_area_digunakan)}")
print(f"Jumlah outlier di kolom 'luas_tanah': {len(outliers_luas_tanah)}")
```
